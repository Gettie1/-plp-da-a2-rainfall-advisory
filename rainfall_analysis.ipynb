```python
# Import NumPy for numerical calculations
import numpy as np

# Import month and county information
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
```

```python
# Rainfall data for the first four counties
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]
]

# Raw Machakos data (stored as strings)
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]
```

```python
def clean_machakos(raw_values, month_names):

    # Store valid numerical readings
    valid_values = []

    # First pass: identify valid readings
    for value in raw_values:
        try:
            number = int(value.strip())
            valid_values.append(number)

        except ValueError:
            continue

    # Calculate the average of valid readings
    average = round(sum(valid_values) / len(valid_values))

    # Store cleaned readings
    cleaned_values = []

    # Store months that require imputation
    imputed_months = []

    # Second pass: clean and replace invalid values
    for i, value in enumerate(raw_values):
        try:
            number = int(value.strip())
            cleaned_values.append(number)

        except ValueError:
            cleaned_values.append(average)
            imputed_months.append(month_names[i])

    return cleaned_values, imputed_months, average


# Clean the Machakos data
machakos_clean, imputed_months, imputed_average = clean_machakos(
    machakos_raw, months
)

# Display the results
print("Imputed average:", imputed_average)
print("Months imputed:", imputed_months)
```

```python
# Combine all five counties
all_rainfall = clean_rows + [machakos_clean]

# Convert the combined data into a NumPy array
rainfall = np.array(all_rainfall)

# Display the array
print("Rainfall data:")
print(rainfall)

# Display array information
print("Array shape:", rainfall.shape)
print("Array data type:", rainfall.dtype)
```

```python
# Calculate the total rainfall for each county
annual_totals = np.sum(rainfall, axis=1)

# Display county totals using zip()
print("Annual Rainfall Totals:")

for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")
```

```python
# Calculate average rainfall for each month
monthly_averages = np.mean(rainfall, axis=0)

# Display monthly averages
print("Average Monthly Rainfall Across All Counties:")

for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python
# Find the index of the maximum rainfall in each county
wettest_indices = np.argmax(rainfall, axis=1)

# Convert indices into month names
wettest_months = [months[index] for index in wettest_indices]

# Pair counties with their wettest months using zip()
print("Wettest Month Per County:")

for county, month in zip(counties, wettest_months):
    print(f"{county}: {month}")
```

```python
# Create a Boolean mask for rainfall below 50 mm
dry_mask = rainfall < 50

# Count True values across each county's row
dry_month_counts = np.sum(dry_mask, axis=1)

# Display the results
print("Number of Dry Months Per County:")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```

```python
# Add each month to the following month
consecutive_rainfall = rainfall[:, :-1] + rainfall[:, 1:]

# Display combined rainfall for consecutive months
print("Combined Rainfall for Consecutive Months:")
print(consecutive_rainfall)
```

```python
# Find the index of the highest consecutive rainfall
planting_indices = np.argmax(consecutive_rainfall, axis=1)

# Convert the indices into two-month planting windows
planting_windows = [
    f"{months[index]}-{months[index + 1]}"
    for index in planting_indices
]

# Find the maximum combined rainfall for each county
highest_window_rainfall = np.max(consecutive_rainfall, axis=1)

# Display planting recommendations
print("Recommended Planting Windows:")

for county, window, total in zip(
    counties, planting_windows, highest_window_rainfall
):
    print(f"{county}: {window} ({total} mm)")
```

```python
import csv

# Define the CSV column headings
headers = [
    "county",
    "annual_total_mm",
    "wettest_month",
    "dry_months",
    "planting_window"
]

# Open the CSV file for writing
with open("rainfall_summary.csv", "w", newline="") as file:

    # Create a CSV writer
    writer = csv.writer(file)

    # Write the column headings
    writer.writerow(headers)

    # Write one row for each county
    for row in zip(
        counties,
        annual_totals,
        wettest_months,
        dry_month_counts,
        planting_windows
    ):
        writer.writerow(row)

print("rainfall_summary.csv created successfully!")
```
